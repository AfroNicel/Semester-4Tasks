```python
import dispy

#nodes = ['192.168.2.*','192.168.1.*']
nodes = ['0.0.0.0','192.168.1.*']

def add_number(num):
    result = num + 1
    return result

cluster = dispy.JobCluster(add_number, nodes=nodes, loglevel=dispy.logger.INFO)

jobs = []
id=1
for n in range(1,100):
    job = cluster.submit(n)
    job.id = id # Associate an ID to the job
    jobs.append(job)
    id += 1   # Next job

print( "Waiting..." )
cluster.wait()

for job in jobs:
    print(f"Job #{job.id} result is: {job.result}")

cluster.print_status()
```

```python
# function 'compute' is distributed and executed with arguments
# supplied with 'cluster.submit' below
def compute(n):
    import time, socket
    time.sleep(n)
    host = socket.gethostname()
    return (host, n)

#if __name__ == '__main__':
# executed on client only; variables created below, including modules imported,
# are not available in job computations
import dispy, random
# distribute 'compute' to nodes; in this case, 'compute' does not have
# any dependencies to run on nodes
cluster = dispy.JobCluster(compute)

# run 'compute' with 20 random numbers on available CPUs
jobs = []
for i in range(20):
    job = cluster.submit(random.randint(5, 20))
    jobs.append(job)
# cluster.wait() # waits until all jobs finish

for job in jobs:
    host, n = job() # waits for job to finish and returns results
    print('%s executed job %s at %s with %s' % (host, job.id, job.start_time, n))
    # other fields of 'job' that may be useful:
    # job.stdout, job.stderr, job.exception, job.ip_addr, job.end_time

cluster.print_status()  # shows which nodes executed how many jobs etc.
```

# Tugas Praktikum: Pemrograman di Klaster

*   Buat dua fungsi berbeda (f1, f2) yang setiap fungsinya mengembalikan nilai numerik (integer/float)
*   Buat job untuk menjalankan 2 fungsi tersebut masing-masing 100 jobs
*   Tunggu semua job selesai, simpan hasil ke list terpisah untuk setiap fungsi
*   Jumlahkan semua job di setiap list dari fungsi tersebut
*   Tampilkan hasil penjumlahan sebagai nilai akhir
*   Buat laporan dalam PDF berisi kode dan penjelasannya





```python
import dispy

def f1(x):
    return x + 5

def f2(x):
    return x * 2

cluster1 = dispy.JobCluster(f1)
cluster2 = dispy.JobCluster(f2)

jobs1 = []
jobs2 = []

for i in range(1, 101):
    jobs1.append(cluster1.submit(i))
    jobs2.append(cluster2.submit(i))

cluster1.wait()
cluster2.wait()

hasil1 = [job() for job in jobs1]
hasil2 = [job() for job in jobs2]

total1 = sum(hasil1)
total2 = sum(hasil2)

print("Total f1 =", total1)
print("Total f2 =", total2)
```
